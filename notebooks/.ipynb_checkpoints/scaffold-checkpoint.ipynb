from the ‘traces’ dataset, select to_address, from_address, transaction_hash, and call_type values for every trace_id that appears over the observation window; download table locally as 'filtered-traces'

```python
%%bigquery filtered-traces

SELECT
    trace_id,
    transaction_hash,
    from_address,
    to_address,
    trace_type,
    call_type,
    `input`,
    block_number,

FROM `bigquery-public-data.crypto_ethereum.traces`
WHERE trace_type IN ('call', 'create')
    AND to_address IS NOT NULL
    AND `bigquery-public-data.crypto_ethereum.traces`.block_timestamp > TIMESTAMP_SUB(CURRENT_TIMESTAMP(), INTERVAL 100 DAY)
```

from the ‘contracts’ dataset: extract address, bytecode, and funtion_sighashes columns; download table locally as 'contracts'.

```python
%%bigquery contracts

SELECT
    address,
    bytecode,
    function_sighashes,

 FROM `bigquery-public-data.crypto_ethereum.contracts`
 WHERE `bigquery-public-data.crypto_ethereum.contracts`.bytecode <> '0x'
 GROUP BY address, bytecode, function_sighashes
```

```python
%config SqlMagic.autopandas = True
%config SqlMagic.feedback = False
%config SqlMagic.displayicon = False
import duckdb
import pandas as pd

%load_ext sql
%sql duckdb:///7999_census.db
```

```python
%%sql

CREATE TABLE filtered-traces AS SELECT * FROM filtered-traces;

CREATE TABLE contracts AS SELECT * FROM contracts;
```

limit from_address and to_address entries in filtered-traces based on their presence in the contracts dataset; note their bytecode and function selectors, and save as 'filtered-contracts'.

```python
%%sql << filtered-contracts

SELECT DISTINCT
    address,
    bytecode,
    function_sighashes,
FROM `contracts`
WHERE address IN (
  SELECT from_address
  FROM `filtered-traces`
  UNION DISTINCT
  SELECT to_address
  FROM `filtered-traces`
      WHERE to_address IS NOT NULL
)
```

proxy resolution using call_type field of filtered-traces

```python
%%sql << potential-proxies

SELECT
    trace_id,
    from_address AS potential_proxy,
    to_address AS potential_implementation,
    SUBSTR(`input`, 1, 10) AS selector,
    trace_type,
    call_type,

FROM `filtered-traces`
WHERE call_type IN ('delegatecall', 'callcode')

#check if any of them were upgraded within the window by using this cell as a subquery
```

check collected bytecode sanity by checking if any contracts were redeployed within the observation window

```python
%%sql << redeployed-contracts

SELECT to_address,
COUNT (*) AS create_count
FROM `filtered-traces`
WHERE trace_type = 'create'
GROUP BY to_address
ORDER BY create_count DESC

#redeploys within the window may lead to misattribution of prevalence/usage metrics, so it's probably better to exclude flagged address' metrics entirely
```

total activity counts for each to_address in the filtered-traces table; denominators for downstream pattern prevalence/usage analysis

```python
%%sql << per-address-activity

SELECT
  DISTINCT to_address AS address,
  COUNT (DISTINCT transaction_hash) AS unique_txn_count,
  COUNT (DISTINCT from_address) AS unique_caller_count,
  COUNT (trace_id) AS call_count

FROM `filtered-traces`
WHERE to_address IS NOT NULL --maybe change to the burn address value
    AND trace_type = 'call'

GROUP BY to_address
ORDER BY unique_caller_count DESC, unique_txn_count DESC, call_count DESC;
```

rolled-up activity counts per unique bytecode. the bytecode column of this cell should be fed into the chosen disassembler for phase 2

```python
%%sql << per-bytecode-activity

SELECT
  bytecode,
  COUNT(DISTINCT address) AS active_deployment_count,
  SUM(unique_txn_count) AS total_txn_count,
  SUM(unique_caller_count) AS total_caller_count, #sum of each deployment's unique callers
  SUM(call_count) AS total_call_count
FROM per-address-activity
JOIN filtered-contracts
  ON address
GROUP BY bytecode
ORDER BY active_deployment_count DESC, total_txn_count DESC, total_caller_count DESC, total_call_count DESC;
```
